```python
CATALOG = "training_lab_classic"
SCHEMA = "cold_chain"
VOLUME = "volume1"

NS = f"{CATALOG}.{SCHEMA}"
BASE = f"/Volumes/{CATALOG}/{SCHEMA}/{VOLUME}"

SENSOR_STAGING = f"{BASE}/staging/sensor_events"
SENSOR_LANDING = f"{BASE}/landing/sensor_events"
SENSOR_CHECKPOINT = f"{BASE}/checkpoints/bronze_sensor_events"
SENSOR_SCHEMA_LOCATION = f"{BASE}/schemas/bronze_sensor_events"

sensor_tables = [
    f"{NS}.bronze_sensor_events",
    f"{NS}.silver_sensor_events",
    f"{NS}.quarantine_sensor_events",
    f"{NS}.silver_sensor_enriched",
    f"{NS}.shipment_temperature_windows",
    f"{NS}.shipment_health",
]

sensor_paths = [
    SENSOR_STAGING,
    SENSOR_LANDING,
    SENSOR_CHECKPOINT,
    SENSOR_SCHEMA_LOCATION,
]

print("Tables scheduled for reset:")
for table_name in sensor_tables:
    print(" -", table_name)

print()

print("Paths scheduled for reset:")
for path in sensor_paths:
    print(" -", path)
```

```python
for table_name in sensor_tables:
    spark.sql(
        f"DROP TABLE IF EXISTS {table_name}"
    )

    print(
        "Dropped table:",
        table_name
    )

for path in sensor_paths:
    try:
        dbutils.fs.rm(
            path,
            recurse=True
        )

        print(
            "Cleared path:",
            path
        )

    except Exception as error:
        print(
            "Path already absent:",
            path
        )

dbutils.fs.mkdirs(SENSOR_STAGING)
dbutils.fs.mkdirs(SENSOR_LANDING)

print()
print("Clean sensor demo state initialized.")
```
