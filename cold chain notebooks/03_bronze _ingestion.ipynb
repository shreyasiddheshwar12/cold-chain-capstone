```python
# Configures the Bronze ingestion notebook, namespace,
# paths, timezone, and shared pipeline run identifier.

from pyspark.sql import functions as F
from pyspark.sql.types import *
import uuid


spark.conf.set(
    "spark.sql.session.timeZone",
    "UTC"
)


# ---------------------------------------------------------
# Namespace configuration
# ---------------------------------------------------------

CATALOG = "training_lab_classic"
SCHEMA = "cold_chain"
VOLUME = "volume1"

NS = f"{CATALOG}.{SCHEMA}"

BASE = (
    f"/Volumes/{CATALOG}/"
    f"{SCHEMA}/{VOLUME}"
)


# ---------------------------------------------------------
# Source paths
# ---------------------------------------------------------

SHIPMENTS_PATH = (
    f"{BASE}/landing/shipments"
)
PRODUCTS_PATH = (
    f"{BASE}/landing/product_limits"
)
DELIVERIES_PATH = (
    f"{BASE}/landing/delivery_events"
)

SENSOR_LANDING = (
    f"{BASE}/landing/sensor_events"
)
SENSOR_STAGING = (
    f"{BASE}/staging/sensor_events"
)


# ---------------------------------------------------------
# Auto Loader state paths
# ---------------------------------------------------------

SENSOR_CHECKPOINT = (
    f"{BASE}/checkpoints/"
    f"bronze_sensor_events"
)
SENSOR_SCHEMA_LOCATION = (
    f"{BASE}/schemas/"
    f"bronze_sensor_events"
)

# ---------------------------------------------------------
# Bronze table names
# ---------------------------------------------------------

BRONZE_PRODUCTS_TABLE = (
    f"{NS}.bronze_product_limits"
)
BRONZE_SHIPMENTS_TABLE = (
    f"{NS}.bronze_shipments"
)
BRONZE_DELIVERIES_TABLE = (
    f"{NS}.bronze_delivery_events"
)
BRONZE_SENSOR_TABLE = (
    f"{NS}.bronze_sensor_events"
)

# ---------------------------------------------------------
# Shared pipeline Run ID
# ---------------------------------------------------------

# The Databricks Job supplies run_id={{job.run_id}}.
# The fallback is used only for direct manual notebook runs.

dbutils.widgets.text(
    "run_id",
    ""
)

RUN_ID_INPUT = (
    dbutils.widgets.get("run_id")
    .strip()
)

RUN_ID = (
    RUN_ID_INPUT
    if RUN_ID_INPUT
    else f"manual_{uuid.uuid4()}"
)

assert RUN_ID, (
    "Pipeline run_id could not be resolved"
)


# ---------------------------------------------------------
# Configuration summary
# ---------------------------------------------------------

print("==============================================")
print("BRONZE INGESTION CONFIGURATION")
print("==============================================")

print("Namespace       :", NS)
print("Base path       :", BASE)
print("Pipeline Run ID :", RUN_ID)
print("Sensor staging  :", SENSOR_STAGING)
print("Sensor landing  :", SENSOR_LANDING)
print("Checkpoint      :", SENSOR_CHECKPOINT)
print("Schema location :", SENSOR_SCHEMA_LOCATION)
```

```python
# Defines explicit schemas for all Bronze source datasets.

product_schema = StructType([
    StructField("product_id", StringType(), True),
    StructField("min_temperature", DoubleType(), True),
    StructField("max_temperature", DoubleType(), True)
])

shipment_schema = StructType([
    StructField("shipment_id", StringType(), True),
    StructField("product_id", StringType(), True),
    StructField("vehicle_id", StringType(), True),
    StructField("origin", StringType(), True),
    StructField("destination", StringType(), True),
    StructField("dispatch_time", TimestampType(), True),
    StructField("expected_arrival", TimestampType(), True)
])

delivery_schema = StructType([
    StructField("event_id", StringType(), True),
    StructField("shipment_id", StringType(), True),
    StructField("status", StringType(), True),
    StructField("event_time", TimestampType(), True)
])

sensor_schema = StructType([
    StructField(
        "event_id",
        StringType(),
        True
    ),
    StructField(
        "shipment_id",
        StringType(),
        True
    ),
    StructField(
        "sensor_id",
        StringType(),
        True
    ),
    StructField(
        "event_time",
        StringType(),
        True
    ),
    StructField(
        "temperature_c",
        StringType(),
        True
    ),
    StructField(
        "batch_id",
        StringType(),
        True
    ),
    StructField(
        "load_mode",
        StringType(),
        True
    )
])

print("Bronze source schemas initialized")
```

```python
PRODUCTS_PATH = (
    "/Volumes/training_lab_classic/"
    "cold_chain/volume1/"
    "landing/product_limits"
)

# Recreate the DataFrame from the current physical files.
products_raw = (
    spark.read
    .option(
        "header",
        "true"
    )
    .option(
        "inferSchema",
        "true"
    )
    .csv(
        PRODUCTS_PATH
    )
)

print(
    "Products rows:",
    products_raw.count()
)

display(
    products_raw
)
```

```python
display(products_raw)
```

```python
# Adds ingestion metadata to product records and writes the Bronze product Delta table.

products_bronze = (
    products_raw
    .withColumn(
        "_source_file",
        F.col("_metadata.file_path")
    )
    .withColumn(
        "_processing_time",
        F.current_timestamp()
    )
    .withColumn(
        "_run_id",
        F.lit(RUN_ID)
    )
)

(
    products_bronze.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(BRONZE_PRODUCTS_TABLE)
)

print("Bronze product table written successfully")
print(
    "Bronze product rows:",
    spark.table(BRONZE_PRODUCTS_TABLE).count()
)

display(
    spark.table(BRONZE_PRODUCTS_TABLE)
)
```

```python
# Reads shipment source files from the landing directory using an explicit schema.

shipments_raw = (
    spark.read
    .option("header", True)
    .schema(shipment_schema)
    .csv(SHIPMENTS_PATH)
)

display(shipments_raw)

print("Shipment source rows:", shipments_raw.count())
```

```python
# Adds ingestion metadata to shipment records and writes the Bronze shipment Delta table.

shipments_bronze = (
    shipments_raw
    .withColumn(
        "_source_file",
        F.col("_metadata.file_path")
    )
    .withColumn(
        "_processing_time",
        F.current_timestamp()
    )
    .withColumn(
        "_run_id",
        F.lit(RUN_ID)
    )
)

(
    shipments_bronze.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(BRONZE_SHIPMENTS_TABLE)
)

print("Bronze shipment table written successfully")
print(
    "Bronze shipment rows:",
    spark.table(BRONZE_SHIPMENTS_TABLE).count()
)

display(
    spark.table(BRONZE_SHIPMENTS_TABLE)
)
```

```python
deliveries_raw = (
    spark.read
    .schema(delivery_schema)
    .json(DELIVERIES_PATH)
)

display(deliveries_raw)
```

```python
# Adds ingestion metadata to delivery events and writes the Bronze delivery Delta table.

deliveries_bronze = (
    deliveries_raw
    .withColumn(
        "_source_file",
        F.col("_metadata.file_path")
    )
    .withColumn(
        "_processing_time",
        F.current_timestamp()
    )
    .withColumn(
        "_run_id",
        F.lit(RUN_ID)
    )
)

(
    deliveries_bronze.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(BRONZE_DELIVERIES_TABLE)
)

print("Bronze delivery table written successfully")
print(
    "Bronze delivery rows:",
    spark.table(BRONZE_DELIVERIES_TABLE).count()
)

display(
    spark.table(BRONZE_DELIVERIES_TABLE)
)
```

```python
# Copies all available sensor JSON files from staging to landing without hardcoding batch names.

staging_files = [
    file_info
    for file_info in dbutils.fs.ls(SENSOR_STAGING)
    if file_info.name.endswith(".json")
]

landing_files = {
    file_info.name
    for file_info in dbutils.fs.ls(SENSOR_LANDING)
    if file_info.name.endswith(".json")
}

copied_files = 0
skipped_files = 0

for file_info in staging_files:
    if file_info.name not in landing_files:
        source_path = file_info.path
        destination_path = (
            f"{SENSOR_LANDING}/{file_info.name}"
        )

        dbutils.fs.cp(
            source_path,
            destination_path
        )

        copied_files += 1

        print(
            "Copied:",
            file_info.name
        )
    else:
        skipped_files += 1

        print(
            "Already in landing:",
            file_info.name
        )

print()
print("Files copied :", copied_files)
print("Files skipped:", skipped_files)

assert (
    copied_files + skipped_files
    == len(staging_files)
), (
    "Staging-file reconciliation failed"
)

print(
    "Total staging files:",
    len(staging_files)
)

print(
    "New files copied:",
    copied_files
)

print(
    "Existing files skipped:",
    skipped_files
)
```

```python
# Displays all JSON sensor files currently available in staging and landing.

staging_sensor_files = [
    file_info
    for file_info in dbutils.fs.ls(SENSOR_STAGING)
    if file_info.name.endswith(".json")
]

landing_sensor_files = [
    file_info
    for file_info in dbutils.fs.ls(SENSOR_LANDING)
    if file_info.name.endswith(".json")
]

print("STAGING FILES")
display(
    spark.createDataFrame(
        [
            (
                file_info.name,
                file_info.path,
                file_info.size
            )
            for file_info in staging_sensor_files
        ],
        [
            "file_name",
            "file_path",
            "size_bytes"
        ]
    )
)

print("LANDING FILES")
display(
    spark.createDataFrame(
        [
            (
                file_info.name,
                file_info.path,
                file_info.size
            )
            for file_info in landing_sensor_files
        ],
        [
            "file_name",
            "file_path",
            "size_bytes"
        ]
    )
)

print(
    "Sensor staging files:",
    len(staging_sensor_files)
)

print(
    "Sensor landing files:",
    len(landing_sensor_files)
)
```

```python
# Creates a single Auto Loader stream for incremental sensor-file ingestion into Bronze.

sensor_stream = (
    spark.readStream
    .format("cloudFiles")
    .option(
        "cloudFiles.format",
        "json"
    )
    .option(
        "cloudFiles.schemaLocation",
        SENSOR_SCHEMA_LOCATION
    )
    .option(
        "cloudFiles.schemaEvolutionMode",
        "rescue"
    )
    .option(
        "rescuedDataColumn",
        "_rescued_data"
    )
    .schema(sensor_schema)
    .load(SENSOR_LANDING)
    .withColumn(
        "_source_file",
        F.col("_metadata.file_path")
    )
    .withColumn(
        "_source_file_name",
        F.col("_metadata.file_name")
    )
    .withColumn(
        "_source_file_size",
        F.col("_metadata.file_size")
    )
    .withColumn(
        "_source_file_modified_at",
        F.col("_metadata.file_modification_time")
    )
    .withColumn(
        "_processing_time",
        F.current_timestamp()
    )
    .withColumn(
        "_run_id",
        F.lit(RUN_ID)
    )
)

print("Sensor Auto Loader stream initialized")
```

```python
# Ingests only files not previously tracked by the Auto Loader
# checkpoint and appends those records into Bronze.

sensor_stream = (
    spark.readStream
    .format("cloudFiles")
    .option(
        "cloudFiles.format",
        "json"
    )
    .option(
        "cloudFiles.schemaLocation",
        SENSOR_SCHEMA_LOCATION
    )
    .option(
        "cloudFiles.schemaEvolutionMode",
        "rescue"
    )
    .option(
        "rescuedDataColumn",
        "_rescued_data"
    )
    .schema(sensor_schema)
    .load(SENSOR_LANDING)
    .withColumn(
        "_source_file",
        F.col("_metadata.file_path")
    )
    .withColumn(
        "_source_file_name",
        F.col("_metadata.file_name")
    )
    .withColumn(
        "_source_file_size",
        F.col("_metadata.file_size")
    )
    .withColumn(
        "_source_file_modified_at",
        F.col(
            "_metadata.file_modification_time"
        )
    )
    .withColumn(
        "_processing_time",
        F.current_timestamp()
    )
    .withColumn(
        "_run_id",
        F.lit(RUN_ID)
    )
)

sensor_query = (
    sensor_stream.writeStream
    .format("delta")
    .outputMode("append")
    .option(
        "checkpointLocation",
        SENSOR_CHECKPOINT
    )
    .option(
        "mergeSchema",
        "true"
    )
    .trigger(
        availableNow=True
    )
    .toTable(
        BRONZE_SENSOR_TABLE
    )
)

sensor_query.awaitTermination()

print(
    "Bronze sensor ingestion completed"
)

print(
    "Bronze sensor rows:",
    spark.table(
        BRONZE_SENSOR_TABLE
    ).count()
)
```

```python
# Profiles raw Bronze sensor data using only columns guaranteed by the sensor schema.

bronze_sensor = spark.table(
    BRONZE_SENSOR_TABLE
)

bronze_quality_profile = (
    bronze_sensor
    .select(
        F.count("*").alias(
            "total_rows"
        ),
        F.sum(
            F.when(
                F.col("event_id").isNull()
                | (F.trim(F.col("event_id")) == ""),
                1
            ).otherwise(0)
        ).alias(
            "missing_event_ids"
        ),
        F.sum(
            F.when(
                F.col("shipment_id").isNull()
                | (F.trim(F.col("shipment_id")) == ""),
                1
            ).otherwise(0)
        ).alias(
            "missing_shipment_ids"
        ),
        F.sum(
            F.when(
                F.col("sensor_id").isNull()
                | (F.trim(F.col("sensor_id")) == ""),
                1
            ).otherwise(0)
        ).alias(
            "missing_sensor_ids"
        ),
        F.sum(
            F.when(
                F.expr(
                    "try_cast(temperature_c AS DOUBLE)"
                ).isNull()
                & F.col("temperature_c").isNotNull(),
                1
            ).otherwise(0)
        ).alias(
            "nonnumeric_temperature_values"
        ),
        F.sum(
            F.when(
                F.expr(
                    "try_cast(event_time AS TIMESTAMP)"
                ).isNull()
                & F.col("event_time").isNotNull(),
                1
            ).otherwise(0)
        ).alias(
            "invalid_event_time_values"
        )
    )
)

display(bronze_quality_profile)

print("Bronze sensor profiling completed.")
```

### testing that rerun dosent add same entries

```python
# Performs final structural checks for the Bronze tables created by this ingestion pipeline.

required_tables = [
    BRONZE_PRODUCTS_TABLE,
    BRONZE_SHIPMENTS_TABLE,
    BRONZE_DELIVERIES_TABLE,
    BRONZE_SENSOR_TABLE
]

missing_tables = [
    table_name
    for table_name in required_tables
    if not spark.catalog.tableExists(
        table_name
    )
]

assert len(missing_tables) == 0, (
    "Missing Bronze tables: "
    + ", ".join(missing_tables)
)

product_bronze_count = (
    spark.table(
        BRONZE_PRODUCTS_TABLE
    ).count()
)

shipment_bronze_count = (
    spark.table(
        BRONZE_SHIPMENTS_TABLE
    ).count()
)

delivery_bronze_count = (
    spark.table(
        BRONZE_DELIVERIES_TABLE
    ).count()
)

sensor_bronze_count = (
    spark.table(
        BRONZE_SENSOR_TABLE
    ).count()
)

print("==============================================")
print("FINAL BRONZE READINESS REPORT")
print("==============================================")

print(
    "Products Bronze   :",
    product_bronze_count
)

print(
    "Shipments Bronze  :",
    shipment_bronze_count
)

print(
    "Deliveries Bronze :",
    delivery_bronze_count
)

print(
    "Sensors Bronze    :",
    sensor_bronze_count
)

print(
    "Required tables   :",
    len(required_tables)
)

print(
    "Missing tables    :",
    len(missing_tables)
)

print(
    "Run ID            :",
    RUN_ID
)

bronze_sensor = spark.table(
    BRONZE_SENSOR_TABLE
)

required_incremental_columns = {
    "batch_id",
    "load_mode",
    "_source_file_name",
    "_processing_time",
    "_run_id"
}

missing_incremental_columns = (
    required_incremental_columns
    - set(bronze_sensor.columns)
)

assert not missing_incremental_columns, (
    "Missing incremental metadata columns: "
    + ", ".join(
        sorted(missing_incremental_columns)
    )
)

batch_summary = (
    bronze_sensor
    .groupBy(
        "load_mode",
        "batch_id",
        "_source_file_name"
    )
    .agg(
        F.count("*").alias(
            "row_count"
        ),
        F.min(
            "_processing_time"
        ).alias(
            "first_processed_at"
        ),
        F.max(
            "_processing_time"
        ).alias(
            "last_processed_at"
        )
    )
    .orderBy(
        "load_mode",
        "batch_id",
        "_source_file_name"
    )
)

print("==============================================")
print("BRONZE INCREMENTAL BATCH SUMMARY")
print("==============================================")

display(batch_summary)

print()
print("Bronze ingestion readiness checks passed.")
```
